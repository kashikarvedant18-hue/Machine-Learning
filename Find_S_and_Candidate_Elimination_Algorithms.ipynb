```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
data= pd.read_csv('enjoy.csv')
data.head()
```

```python
from typing import dataclass_transform
x= data.iloc[:,:-1].values

y= data.iloc[:,-1].values
```

```python
def find_s(x, y):
    hypothesis = None

    print("Initial Hypothesis:")
    print(hypothesis)

    for i in range(len(x)):

        if y[i] == "Yes":
            if hypothesis is None:
                hypothesis = x[i].copy()
            else:
                for j in range(len(hypothesis)):
                    if hypothesis[j] != x[i][j]:
                        hypothesis[j] = '?'

        print(f"\nFinal Hypothesis : {i + 1}")
        print(hypothesis)

    return hypothesis


final_hypothesis = find_s(x, y)

print("\nFinal Hypothesis:")
print(final_hypothesis)
```

```python
def candidate_elimination(concepts, target):

    specific_h = concepts[0].copy()
    general_h = [["?" for i in range(len(specific_h))]]

    print("\nInitial S :", specific_h)
    print("Initial G :", general_h)

    for i in range(len(concepts)):

        example = concepts[i]

        # Positive example
        if target[i] == "Yes":

            for j in range(len(specific_h)):
                if specific_h[j] != example[j]:
                    specific_h[j] = '?'

            general_h = [
                h for h in general_h
                if all(h[j] == '?' or h[j] == example[j]
                       for j in range(len(h)))
            ]

        # Negative example
        else:

            new_general_h = []

            for h in general_h:

                if all(h[j] == '?' or h[j] == example[j]
                       for j in range(len(h))):

                    for j in range(len(specific_h)):

                        if specific_h[j] != '?' and specific_h[j] != example[j]:

                            new_h = h.copy()
                            new_h[j] = specific_h[j]

                            if new_h not in new_general_h:
                                new_general_h.append(new_h)

                else:
                    new_general_h.append(h)

            general_h = new_general_h

        print("\nExample", i + 1)
        print("S =", specific_h)
        print("G =", general_h)

    return specific_h, general_h


# Run Candidate Elimination
S, G = candidate_elimination(x, y)

print("\nFinal Specific:")
print(S)

print("\nFinal General:")
print(G)
```

```python
import pandas as pd

data = pd.read_csv("enjoy.csv")

X = data.iloc[:, :-1].values
y = data.iloc[:, -1].values



def find_s(x, y):

    hypothesis = None

    print("========== FIND-S ALGORITHM ==========")

    print("\nInitial Hypothesis:")
    print(hypothesis)

    for i in range(len(x)):

        if str(y[i]).strip().lower() == "yes":

            if hypothesis is None:
                hypothesis = x[i].copy()

            else:
                for j in range(len(hypothesis)):

                    if hypothesis[j] != x[i][j]:
                        hypothesis[j] = '?'

        print("\nExample", i + 1)
        print("Hypothesis:", hypothesis)

    return hypothesis



def candidate_elimination(concepts, target):

    specific_h = concepts[0].copy()

    general_h = [
        ["?" for i in range(len(specific_h))]
    ]

    print("\n========== CANDIDATE ELIMINATION ==========")

    print("\nInitial S:")
    print(specific_h)

    print("\nInitial G:")
    print(general_h)

    for i in range(len(concepts)):

        example = concepts[i]

        # Positive example
        if str(target[i]).strip().lower() == "yes":

            for j in range(len(specific_h)):

                if specific_h[j] != example[j]:
                    specific_h[j] = '?'

        # Negative example
        else:

            new_general_h = []

            for h in general_h:

                for j in range(len(specific_h)):

                    if h[j] == '?' and specific_h[j] != '?':

                        new_h = h.copy()
                        new_h[j] = specific_h[j]

                        if new_h not in new_general_h:
                            new_general_h.append(new_h)

            if new_general_h:
                general_h = new_general_h

        print("\nExample", i + 1)
        print("S =", specific_h)
        print("G =", general_h)

    return specific_h, general_h



find_s_hypothesis = find_s(X, y)



S, G = candidate_elimination(X, y)



print("       OUTPUT 4 - HYPOTHESIS COMPARISON")

print("\nFind-S Hypothesis:")
print(find_s_hypothesis)

print("\nCandidate Elimination:")
print("Specific Boundary (S):")
print(S)

print("\nGeneral Boundary (G):")
print(G)


print("       OUTPUT 5 - FINAL RESULT")

print("\nFinal Find-S Hypothesis:")
print(find_s_hypothesis)

print("\nFinal Candidate Elimination Specific Boundary:")
print(S)

print("\nFinal Candidate Elimination General Boundary:")
print(G)

print("        PROGRAM EXECUTED SUCCESSFULLY")
```
